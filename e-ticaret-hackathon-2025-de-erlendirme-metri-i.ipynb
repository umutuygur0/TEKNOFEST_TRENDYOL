```python
import pandas as pd

DATA_PATH = "/kaggle/input/trendyol-e-ticaret-hackathonu-2025-kaggle/data/"

import sys
sys.path.append(DATA_PATH)

from trendyol_metric_group_auc import score
```

```python
solution_df = pd.DataFrame({
    "session_id": ["s1", "s2"],
    "ordered_items": ["c1", "c3"],
    "clicked_items": ["c1 c2", "c3 c4"],
    "all_items": ["c1 c2 c3 c4 c5 c6", "c3 c4 c5 c6 c7 c8"]
})
```

```python
# Oturum içerisinde click ve order var, tahminler tam doğru sırada

submission_df = pd.DataFrame({
    "session_id": ["s1", "s2"],
    "prediction": ["c1 c2 c5 c3 c4 c6", "c3 c4 c5 c6 c7 c8"]
})

score(solution_df, submission_df, "session_id")
```
